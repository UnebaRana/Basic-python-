# Linear Regression on Titanic Dataset

This notebook trains a **Linear Regression** model using the provided Titanic training dataset.

### Objective
Predict the `Survived` value using passenger-related features.

**Note:** `Survived` is a binary 0/1 target, so Logistic Regression is normally more suitable for classification. This notebook uses **Linear Regression** specifically for the requested task.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Step 2: Load dataset

train = pd.read_csv("train(1).csv")

print("Train Shape:", train.shape)
```

```python
# Step 3: Display first five rows

train.head()
```

```python
# Step 4: Check dataset information

print("===== DATASET INFORMATION =====")
train.info()

print("\n===== DATASET SHAPE =====")
print("Rows:", train.shape[0])
print("Columns:", train.shape[1])

print("\n===== COLUMN NAMES =====")
print(train.columns.tolist())

print("\n===== FIRST 5 ROWS =====")
display(train.head())
```

```python
# Step 5: Check missing values

print("===== MISSING VALUES =====")
missing_values = train.isnull().sum()
print(missing_values)

print("\n===== MISSING VALUES PERCENTAGE =====")
missing_percentage = (train.isnull().sum() / len(train)) * 100
print(missing_percentage.round(2))
```

```python
# Step 6: Check target variable

print("===== SURVIVED VALUE COUNTS =====")
print(train["Survived"].value_counts())

print("\n===== SURVIVED VALUE PERCENTAGE =====")
print((train["Survived"].value_counts(normalize=True) * 100).round(2))
```

```python
# Target distribution

plt.figure(figsize=(6, 4))
sns.countplot(data=train, x="Survived")
plt.title("Survival Distribution")
plt.xlabel("Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Step 7: Select features and target

features = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "Embarked"
]

X = train[features].copy()
y = train["Survived"].copy()

print("===== FEATURES =====")
print(X.columns.tolist())

print("\n===== X SHAPE =====")
print(X.shape)

print("\n===== TARGET SHAPE =====")
print(y.shape)

print("\n===== FIRST 5 ROWS OF X =====")
display(X.head())

print("\n===== FIRST 5 VALUES OF y =====")
display(y.head())
```

```python
# Step 8: Handle missing values

X["Age"] = X["Age"].fillna(X["Age"].median())
X["Embarked"] = X["Embarked"].fillna(X["Embarked"].mode()[0])

print("===== MISSING VALUES AFTER HANDLING =====")
print(X.isnull().sum())
```

```python
# Step 9: Convert categorical variables into numerical variables

X = pd.get_dummies(
    X,
    columns=["Sex", "Embarked"],
    drop_first=True,
    dtype=int
)

print("===== COLUMNS AFTER ENCODING =====")
print(X.columns.tolist())

print("\n===== FIRST 5 ROWS =====")
display(X.head())

print("\n===== DATA TYPES =====")
print(X.dtypes)
```

```python
# Step 10: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("===== DATA SPLIT =====")
print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
# Step 11: Feature Scaling

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("===== SCALING COMPLETE =====")
print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape :", X_test_scaled.shape)
```

```python
# Step 12: Train Linear Regression Model

model = LinearRegression()
model.fit(X_train_scaled, y_train)

print("Linear Regression model trained successfully!")
```

```python
# Step 13: Display model coefficients

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

print("===== MODEL COEFFICIENTS =====")
display(coefficients)

print("Intercept:", model.intercept_)
```

```python
# Step 14: Make predictions

y_pred = model.predict(X_test_scaled)

print("First 20 Predictions:")
print(np.round(y_pred[:20], 4))

print("\nFirst 20 Actual Values:")
print(y_test.values[:20])
```

```python
# Step 15: Evaluate the model

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("===== LINEAR REGRESSION MODEL PERFORMANCE =====")
print("Mean Absolute Error (MAE):", round(mae, 4))
print("Mean Squared Error (MSE):", round(mse, 4))
print("Root Mean Squared Error (RMSE):", round(rmse, 4))
print("R2 Score:", round(r2, 4))
```

```python
# Step 16: Actual vs Predicted values

results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

display(results.head(20))
```

```python
# Step 17: Actual vs Predicted graph

plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, alpha=0.6)

min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())
plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.title("Linear Regression - Actual vs Predicted")
plt.xlabel("Actual Survived")
plt.ylabel("Predicted Survived")
plt.grid(True)
plt.show()
```

```python
# Step 18: Prediction distribution

plt.figure(figsize=(7, 5))
plt.hist(y_pred, bins=20)

plt.title("Distribution of Linear Regression Predictions")
plt.xlabel("Predicted Survived Value")
plt.ylabel("Frequency")
plt.show()
```

## Conclusion

The Linear Regression model has been trained on the Titanic dataset.

The model is evaluated using:
- MAE (Mean Absolute Error)
- MSE (Mean Squared Error)
- RMSE (Root Mean Squared Error)
- R² Score

Since `Survived` is a binary 0/1 target, Linear Regression may produce decimal predictions. For a true survival classification problem, Logistic Regression would normally be the appropriate model.
